```
import sys, os, subprocess
sys.path.insert(0, '/content/YuE/src')
from yue2 import YuE2Pipeline

request = {
    'style': "English, cheerful children's nursery rhyme, bright female child vocal, acoustic ukulele and playful xylophone, simple memorable melody, slow clear pronunciation for kids, major key, 96 BPM",
    'lyrics': "[Verse]\nHello, Pip! Hello, Pip!\nHow are you today?\nI am sleepy, yawn, yawn, yawn!\nI am hungry, oh no!\n\n[Chorus]\nHere are cookies, yummy, yummy!\nNow I'm happy, jump up high!\nHow are you? I am happy!\nBye-bye, Pip! Bye-bye, Mia!",
    'cot': 'full',
    'seed': 202601,
}

with YuE2Pipeline.from_pretrained('m-a-p/YuE2-3B', device='cuda', low_vram=True) as pipe:
    song = pipe(**request)
    song.save_artifacts('/content/hello_pip_yue2')

subprocess.check_call('ffmpeg -y -i /content/hello_pip_yue2/audio.flac -b:a 192k /content/hello_pip_yue2.mp3', shell=True)
from google.colab import files
files.download('/content/hello_pip_yue2.mp3')
```
